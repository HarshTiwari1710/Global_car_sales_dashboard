```python
cleanup_map = {
    'electrc': 'electric',
    'petrole': 'petrol',
    'diesal': 'diesel',
    'hybrd': 'hybrid'
}
df_copy['engine_type'] = df_copy['engine_type'].str.lower().str.strip()
df_copy['engine_type'] = df_copy['engine_type'].replace(cleanup_map)
```

```python
def fill_with_mode(series):
    # Returns the mode if it exists, otherwise the original missing value
    mode = series.mode()
    if not mode.empty:
        return series.fillna(mode.iloc[0])
    return series
df_copy['engine_type'] = df_copy.groupby('model')['engine_type'].transform(fill_with_mode)
```

```python
df_copy['engine_type'].isna().sum()
```

```python
country_map = {
    # Australia
    'australia': 'Australia',
    'Australlia': 'Australia',
    'AUSTRALIA': 'Australia',
    
    # Brazil
    'Brasil': 'Brazil',
    'BRAZIL': 'Brazil',
    
    # Canada
    'CANADA': 'Canada',
    
    # China
    'CHINA': 'China',
    
    # France
    'Frnce': 'France',
    'FRANCE': 'France',
    
    # Germany
    'GERMANY': 'Germany',
    'Germny': 'Germany',
    'Germnay': 'Germany',
    
    # India
    'INDIA': 'India',
    
    # Italy
    'ITALY': 'Italy',
    
    # Japan
    'JAPAN': 'Japan',
    
    # Mexico
    'MEXICO': 'Mexico',
    
    # Netherlands
    'NETHERLANDS': 'Netherlands',
    
    # Spain
    'SPAIN': 'Spain',
    
    # USA
    'U.S.A': 'USA',
    'US': 'USA',
    'usa': 'USA'
}

df_copy['country'] = df_copy['country'].str.strip()
df_copy['country'] = df_copy['country'].replace(country_map)
```

```python
segment_map = {
    # SUV
    'S U V': 'SUV',
    'suv': 'SUV',
    'Suv': 'SUV',
    
    # Sedan
    'sedan': 'Sedan',
    'Sedaan': 'Sedan',
    'SEDAN': 'Sedan',
    
    # Electric (Interestingly classified as a segment for some rows)
    'Electrc': 'Electric',
    'electric': 'Electric'
}

# How to apply:
df_copy['segment'] = df_copy['segment'].str.strip()
df_copy['segment'] = df_copy['segment'].replace(segment_map)
```

```python
# replace the null values in the country column with the mode of the country column
df_copy['country']=df_copy['country'].fillna(df_copy['country'].mode()[0], inplace=True)   
```

```python
df_copy['segment'] = df_copy['segment'].fillna(df['segment'].mode()[0])
```

```python
df_copy.isna().sum()
```

```python
df['model'] = df['model'].str.strip()
```

```python
df_copy['price_usd'] = df_copy['price_usd'].replace({r'\$': '', r',': ''}, regex=True)
df_copy['price_usd'] = pd.to_numeric(df_copy['price_usd'], errors='coerce')
```

```python
def fill_with_median(series):
    # Only fill if the group has a valid median
    return series.fillna(series.median())
# Apply the function grouping by 'model'
df_copy['price_usd'] = df_copy.groupby('model')['price_usd'].transform(fill_with_median)
# Fallback: Just in case there's a rare model where ALL prices are missing, 
# fill whatever is left with the median of the entire Brand
df_copy['price_usd'] = df_copy.groupby('brand')['price_usd'].transform(fill_with_median)
```

```python
# 1. Strip the dollar signs and commas, convert to numeric
df_copy['marketing_spend_usd'] = df_copy['marketing_spend_usd'].replace({r'\$': '', r',': '', r'"': ''}, regex=True)
df_copy['marketing_spend_usd'] = pd.to_numeric(df_copy['marketing_spend_usd'], errors='coerce')
```

```python
# 2. Fill missing marketing spend with the median of that specific model
df_copy['marketing_spend_usd'] = df_copy.groupby('model')['marketing_spend_usd'].transform(fill_with_median)
```

```python
# 3. Fallback to Brand median if needed
df_copy['marketing_spend_usd'] = df_copy.groupby('brand')['marketing_spend_usd'].transform(fill_with_median)
```

```python
df_copy['fuel_price_usd'] = pd.to_numeric(df_copy['fuel_price_usd'], errors='coerce')
df_copy['fuel_price_usd'] = df_copy.groupby(['country', 'year'])['fuel_price_usd'].transform(lambda x: x.fillna(x.median()))
df_copy['gdp_growth_percent'] = pd.to_numeric(df_copy['gdp_growth_percent'], errors='coerce')
df_copy['gdp_growth_percent'] = df_copy.groupby(['country', 'year'])['gdp_growth_percent'].transform(lambda x: x.fillna(x.median()))
```

```python
df_copy['dealership_count'] = pd.to_numeric(df_copy['dealership_count'], errors='coerce')
df_copy['dealership_count'] = df_copy.groupby(['brand', 'country'])['dealership_count'].transform(lambda x: x.fillna(x.median()))
```

```python
# 1. Force to numeric
df_copy['units_sold'] = pd.to_numeric(df_copy['units_sold'], errors='coerce')

# 2. Fix negative values by taking the absolute value
df_copy['units_sold'] = df_copy['units_sold'].abs()

# 3. Drop any rows where units_sold is STILL missing
df_copy = df_copy.dropna(subset=['units_sold'])
```

```python
df_copy = df_copy.drop_duplicates()
```

```python
df_copy['brand'] = df_copy['brand'].str.strip().str.title()
```

```python
brand_map = {
    'Bmw': 'BMW',
    'B M W': 'BMW',
    'Audo': 'Audi',
    'Porshe': 'Porsche',
    'Teslla': 'Tesla',
    'Telsla': 'Tesla',
    'Frod': 'Ford',
    'Toyoda': 'Toyota',
    'Toyata': 'Toyota'
}
df_copy['brand'] = df_copy['brand'].replace(brand_map)
```

```python
df_copy['engine_type'] = df_copy['engine_type'].str.strip().str.lower()
```

```python
df_copy['year'] = pd.to_numeric(df_copy['year'], errors='coerce')
df_copy.loc[df_copy['year'] > 2025, 'year'] = np.nan
```

```python
df_copy['month'] = pd.to_numeric(df_copy['month'], errors='coerce')
df_copy.loc[(df_copy['month'] > 12) | (df_copy['month'] < 1), 'month'] = np.nan
```

```python
df_copy['price_usd'] = pd.to_numeric(df_copy['price_usd'], errors='coerce')
df_copy.loc[df_copy['price_usd'] > 500000, 'price_usd'] = 500000
```

```python
df_copy = df_copy.dropna(subset=['dealership_count', 'fuel_price_usd', 'gdp_growth_percent'])
```

```python
df_copy['model'] = df_copy['model'].str.strip()
```

```python
print("Total rows left:", len(df_copy))
print(df_copy.isna().sum())
```

```python
df_copy = df_copy.dropna(subset=['year', 'month'])
# Verify it's perfect!
print(df_copy.isna().sum())
```

```python
if 'Unnamed: 0' in df_copy.columns:
    df_copy = df_copy.drop(columns=['Unnamed: 0'])
```

```python
df_copy = df_copy[df_copy['price_usd'] < 500000]
```

```python
df_copy['dealership_count'] = df_copy['dealership_count'].replace(0, np.nan)
df_copy['dealership_count'] = df_copy.groupby('brand')['dealership_count'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df_copy['engine_type'] = df_copy['engine_type'].str.title()
```

```python
valid_segments = ['Suv', 'Sedan', 'Hatchback', 'Truck', 'Coupe'] # Add any others you have
df_copy['segment'] = df_copy['segment'].str.title().str.strip()
```

```python
true_segments = (
    df_copy[df_copy['segment'].isin(valid_segments)]
    .groupby('model')['segment']
    .agg(lambda x: pd.Series.mode(x)[0] if not pd.Series.mode(x).empty else np.nan)
)
```

```python
df_copy['segment'] = df_copy['model'].map(true_segments)
# Just in case mapping left a few NaNs, drop them
df_copy = df_copy.dropna(subset=['segment'])
```
